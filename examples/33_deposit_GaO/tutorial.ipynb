# Deposit
## Introduction

In this tutorial, we will show you how to use the ``GPUMD`` package to deposit atoms to surface.
For the information related to deposition, please read the [documentation](https://gpumd.org/gpumd/input_parameters/deposit.html). 


### Prepare the `model.xyz` file. 
- Use the ``model.xyz`` file for GaO from Tutorial 16   

```
5262
Lattice="50.0 0.0 0.0 0.0 50.0 0.0 0.0 0.0 100.0" Properties=species:S:1:pos:R:3:vel:R:3:group:I:1 pbc="T T F"
O 22.0574 26.6236 21.3925 0.0 0.0 0.0 2
Ga 44.3672 40.0923 11.42444 0.0 0.0 0.0 0
O 39.6287 2.17761 28.5361 0.0 0.0 0.0 1
Ga 29.5949 13.1155 20.0729 0.0 0.0 0.0 1
O 42.0032 6.22095 11.98315 0.0 0.0 0.0 0
Ga 37.1865 49.3785 34.3591 0.0 0.0 0.0 2
```


### Prepare the `run.in` file. 
- First, we need to identify the elements being deposited, as well as their respective quantities and rates.     
A example ``run.in`` file reads:
```
potential        nep.txt
velocity         300
time_step        0.5

ensemble         heat_nhc 300 100 0 1 2
fix              0
dump_xyz         -1 0 1000 dump.xyz group
deposit          20000 2 60 70 atom 0 20 -0.005 1 20 -0.005
#deposit          20000 2 60 file add_Ga_O.txt -0.005
#deposit          20000 -1 0 file add_Ga_O.txt
run              200000
```

Here are three examples:
1. Add 20 atoms each with indices 0 and 1 in the z-direction between 60 and 70 Å, all with a velocity of 0.005 Å/fs.
2. Add the atoms from the file along the z-axis; their z-axis coordinates and velocities will be reset to 60 Å and -0.005 Å/fs, respectively.
3. Use only the atomic information (coordinates and velocities) from the file; the height settings will be ignored.
